<font color= "red"> 1.1-8 (a) Mostre que a potência de um sinal</font>
$$
x(t)=\sum_{k=m}^n D_k e^{j \omega_k t} \quad \text { é } \quad P_x=\sum_{k=m}^n\left|D_k\right|^2
$$
assumindo que todas as frequências são distintas, ou seja, $\omega_i \neq \omega_k$ para todo $i \neq k$.

(b) Usando o resultado da parte (a), determine a potência de cada um dos sinais do Problema 1.1-5.

#### Letra (a)

Dado o sinal:

$$
x(t) = \sum_{k=m}^n D_k e^{j \omega_k t}
$$

A potência $P_x$ de $x(t)$ pode ser calculada como:

$$
P_x = \lim_{T \to \infty} \frac{1}{T} \int_{-\frac{T}{2}}^{\frac{T}{2}} |x(t)|^2 dt
$$

Substituindo a expressão de $x(t)$:

$$
P_x = \lim_{T \to \infty} \frac{1}{T} \int_{-\frac{T}{2}}^{\frac{T}{2}} \left| \sum_{k=m}^n D_k e^{j \omega_k t} \right|^2 dt
$$

Como as frequências são distintas, os termos exponenciais são ortogonais entre si, então podemos separar o módulo ao quadrado da soma em somas de módulos ao quadrado de cada termo:

$$
P_x = \lim_{T \to \infty} \frac{1}{T} \int_{-\frac{T}{2}}^{\frac{T}{2}} \left| \sum_{k=m}^n D_k e^{j \omega_k t} \right|^2 dt \\
= \lim_{T \to \infty} \frac{1}{T} \int_{-\frac{T}{2}}^{\frac{T}{2}} \sum_{k=m}^n \left| D_k e^{j \omega_k t} \right|^2 dt \\
= \lim_{T \to \infty} \frac{1}{T} \int_{-\frac{T}{2}}^{\frac{T}{2}} \sum_{k=m}^n \left| D_k \right|^2 \left| e^{j \omega_k t} \right|^2 dt \\
= \lim_{T \to \infty} \frac{1}{T} \int_{-\frac{T}{2}}^{\frac{T}{2}} \sum_{k=m}^n \left| D_k \right|^2 dt \\
= \lim_{T \to \infty} \frac{1}{T} \sum_{k=m}^n \left| D_k \right|^2 \int_{-\frac{T}{2}}^{\frac{T}{2}} dt \\
= \lim_{T \to \infty} \frac{1}{T} \sum_{k=m}^n \left| D_k \right|^2 \cdot T \\
= \sum_{k=m}^n \left| D_k \right|^2
$$

Portanto, demonstramos que a potência do sinal $x(t)$ é dada por $P_x = \sum_{k=m}^n \left| D_k \right|^2$, assumindo que todas as frequências são distintas.


#### Letra (b)

A) $ x(t) = 5 + 10 \cos(100t + \frac{\pi}{3}) $

Expressando o sinal na forma complexa de Euler:

$$ x(t) = 5 + 10 \cos(100t + \frac{\pi}{3}) = 5 + 10 \left( \frac{e^{j(100t + \frac{\pi}{3})} + e^{-j(100t + \frac{\pi}{3})}}{2} \right) $$

Agora podemos identificar os coeficientes \( D_k \):

$$ D_0 = 5 $$
$$ D_1 = \frac{10}{2} = 5 $$

Para todos os outros \( k \), os coeficientes são zero. Então, podemos calcular a potência usando a fórmula obtida anteriormente:

$$ P_x = \sum_{k=m}^n \left| D_k \right|^2 = \left| D_0 \right|^2 + \left| D_1 \right|^2 = 5^2 + 5^2 = 25 + 25 = 50 $$

Portanto, a potência do sinal é \( P_x = 50 \).



B) $ x(t) = 10 \cos(100t + \frac{\pi}{3}) + 16 \sin(150t + \frac{\pi}{5}) $

$$ x(t) = 10 \cos(100t + \frac{\pi}{3}) + 16 \sin(150t + \frac{\pi}{5}) $$
$$ = 10 \left( \frac{e^{j(100t + \frac{\pi}{3})} + e^{-j(100t + \frac{\pi}{3})}}{2} \right) + 16j \left( \frac{e^{j(150t + \frac{\pi}{5})} - e^{-j(150t + \frac{\pi}{5})}}{2j} \right) $$



$$ D_1 = \frac{10}{2} = 5 $$
$$ D_2 = \frac{16j}{2j} = 8j $$


$$ P_x = \sum_{k=m}^n \left| D_k \right|^2 = \left| D_1 \right|^2 + \left| D_2 \right|^2 $$
$$ = \left| 5 \right|^2 + \left| 8j \right|^2 = 5^2 + (8j)(-8j) = 25 + 64 = 89 $$

Portanto, a potência do sinal é \( P_x = 89 \).

<font color= "red">1.2-4 Para um sinal de energia $x(t)$ com energia $E_x$, mostre que a energia de qualquer um dos sinais $-x(t), x(-t)$ e $x(t-T)$ é $E_x$. Mostre também que a energia de $x(a t)$ e $x(a t-b)$ é $E_x / a$, mas a energia de $a x(t)$ é $a^2 E_x$. Isto mostra que a inversão temporal e o deslocamento temporal não afetam a energia do sinal. Por outro lado, a compressão no tempo de um sinal ( $a$ $>1$ ) reduz a energia e a expansão no tempo de um sinal $(a<1)$ aumenta a energia. Qual é o efeito na energia do sinal se o sinal for multiplicado por uma constante $a$?</font>


#### Resolução

Para inversão e reflexão do sinal, os sinais negativos em $x$ e $t$, respectivamente, não interferem, pois a energia é calculada a partir do módulo do sinal.

Para o deslocamento, temos:

$ E_{x(t-T)} = \int^\infty_{-\infty} |x(t-T)|^2 dt \\ = \int^\infty_{-\infty} |x(t)|^2 dt \\ = E_{x(t)} $

Para compressão, temos:

$ E_{x(t-T)} = \int^\infty_{-\infty} |x(at)|^2 dt \\ = \frac{1}{|a|} \cdot \int^\infty_{-\infty} |x(t)|^2 dt \\ =  \frac{E_{x(t)}}{|a|} $

Fazendo uma multiplicação por uma constante, temos:

$ E_{x(t-T)} = \int^\infty_{-\infty} |a \cdot x(t)|^2 dt \\ = a^2 \cdot \int^\infty_{-\infty} |x(t)|^2 dt \\ = a^2 \cdot E_{x(t)} $

<font color= "red">1.3-2 Determine se cada uma das seguintes afirmativas é verdadeira ou falsa. Se a afirmativa for falsa, demonstre por prova ou exemplo porque a afirmativa é falsa.</font>

(a) Todo sinal periódico limitado é um sinal de potência.

(b) Todo sinal de potência limitado é um sinal periódico

(c) Se um sinal de energia $x(t)$ possui energia $E$, então a energia de $x(a t)$ é $E / a$. Considere $a$ um número real positivo.

(d) Se um sinal de potência $x(t)$ possui potência $P$, então a potência de $x(a t)$ é $P / a$. Considere $a$ um número real positivo.

#### Resolução:

(a) Falsa: Não é verdade que todo sinal periódico limitado é um sinal de potência. Um contraexemplo pode ser o pulso retangular, que é periódico e limitado, mas possui energia finita, o que o torna um sinal de energia, não um sinal de potência.

(b) Falsa: Nem todo sinal de potência limitado é periódico. Um exemplo é o sinal $ x(t) = e^{-t^2} $, que é limitado, mas não é periódico. Portanto, esta afirmativa é falsa.

(c) Verdadeira: Se um sinal de energia $ x(t) $ possui energia $ E $ , então a energia de $ x(at) $ é $ \frac{E}{|a|} $. Isso ocorre porque uma mudança na escala do tempo afeta a amplitude do sinal, mas não a sua energia total. Se $ a $ for negativo, a mudança de escala temporal também inverte o sinal $ x(at) $, mas isso não afeta a energia total, que é uma quantidade não negativa.

(d) Falsa: Se um sinal de potência $ x(t) $ possui potência $ P $, então a potência de $ x(at) $ é $ \frac{P}{|a|} $. Isso ocorre porque uma mudança na escala do tempo afeta tanto a amplitude quanto a densidade de potência do sinal. Se $ a $ for negativo, a mudança de escala temporal inverte o sinal de $ x(at) $, o que afeta a densidade de potência, mas não afeta a potência total, que é uma quantidade não negativa.
