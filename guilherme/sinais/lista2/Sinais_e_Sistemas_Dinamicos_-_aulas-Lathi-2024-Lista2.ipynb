1.1-1 Determine a energia dos sinais mostrados na Fig. P1.1-1. Comente o efeito na energia da mudança de sinal, deslocamento temporal ou escalamento (dobro) do sinal. Qual é o efeito na energia se o sinal for multiplicado por $k$ ?





**Solução:**

Seja $x(t)$ o sinal da Fig. P1.1-1a. A energia deste sinal é dada por 

$$
E_x=\int_{-\infty}^{\infty} x^2(t) d t
$$

1. Podemos observar que 
$$
E_x=\int_{-\infty}^{\infty} x^2(t) d t = E_{-x}=\int_{-\infty}^{\infty} (-x)^2(t) d t. 
$$
Assim, a energia do sinal não se altera por mudança de sinal.


2. No caso do deslocamento temporal, temos para uma constante $T$ que
$$
E_{x_T}=\int_{-\infty}^{\infty} x^2(t - T) d t
$$
Fazendo uma mundaça de variáveis $\tau = t - T$ e observando que $d\tau = dt$, obtemos 
$$
E_{x_T}=\int_{-\infty}^{\infty} x^2(\tau) d \tau = E_x,
$$
Donde, inferimos que o deslocamento temporal não altera a energia do sinal.


3. Para o escalamento por um fator $a > 0$, temos 
$$
E_{x_a}=\int_{-\infty}^{\infty} x^2(at) d t,
$$
Fazendo uma mudança de variáveis, $\tau = at$, obtemos 
$$
E_{x_a}=\int_{-\infty}^{\infty} x^2(\tau) ad \tau = \frac{1}{a}\int_{-\infty}^{\infty} x^2(\tau) d \tau = \frac{1}{a}E_x.
$$
Observação: para o caso em que $a < 0$, tem-se que 
$$
E_{x_a}=\int_{-\infty}^{\infty} x^2(at) d t = \int_{\infty}^{-\infty} x^2(\tau) \frac{1}{a}d \tau = -\frac{1}{a}E_x.
$$
E assim, podemos enunciar que para um escalamento por um fator $a \neq 0$ temos que
$$
E_{x_a}= \left|\frac{1}{a}\right|E_x.
$$

4. Finalmente, se o sinal for multiplicado por $k$, a energia é multipçlicada por $k^2$, como podemos observar abaixo:
$$
E_{kx}=\int_{-\infty}^{\infty} [kx^2(t)] d t = k^2 E_{x}.
$$
