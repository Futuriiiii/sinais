### Questão 1.

Considere o sistema $S$ tal que $S[g(t)] = \frac{1}{2}\int^{+ \infty}_{-\infty} g(\tau) \left[\delta(t - \tau) + \delta(t +  \tau) \right]d\tau$:

i. descreva o comportamento do sistema S. Qual a saída esperada para um sinal par? e para um sinal ímpar?


Devemos notar que 

$$
\begin{aligned}
S[g(t)] &= \frac{1}{2}\int^{+ \infty}_{-\infty} g(\tau) \left[\delta(t - \tau) + \delta(t +  \tau) \right]d\tau\\
& = \frac{1}{2} \left[g(t)\int^{+ \infty}_{-\infty}  \delta(t - \tau)d\tau + g(-t)\int^{+ \infty}_{-\infty}  \delta(t + \tau)d\tau \right]\\
& = \frac{1}{2} \left[g(t) + g(-t)\right] = g_p(t),
\end{aligned}
$$
ou seja o sistema $S[g(t)]$ fornece como resposta a componente par $g_p(t)$ do sinal de entrada $g(t)$. Se o sinal de entrada for par, a saída é o próprio sinal, tendo se em vista que $g_p(t) = g(t)$. Se o sinal $g(t)$ de entrada for ímpar, temos que $g_p(t) = 0$, isto é, a saída é nula.

ii. O sistema $S$ é linear? Invariante no tempo? Causal? Justifique.

 - Notemos que 
 $$
 \begin{aligned}
 S[\sum_{i} a_i g_i(t)] & = \frac{1}{2} \left[\sum_{i} a_i g_i(t) + \sum_{i} a_i g_i(-t) \right]\\
 & = \sum_{i} a_i\frac{1}{2} \left[ g_i(t) + g_i(-t) \right]\\
 & = \sum_{i} a_i S\left[g_i(t) \right],
 \end{aligned}
 $$
 
 indicando que o sistema $S$ é **linear**.
 
 - O sistema é **variante no tempo**, tendo-se em vista que 
 
 $$
 \begin{aligned}
 S[T[g(t)]] & = S[g(t - T_0)] = \frac{1}{2}\left[g(t - T_0) + g(-t + T_0)\right]\\
 T[S[g(t)]] & = T[\frac{1}{2}\left[g(t) + g(-t)\right]] = \frac{1}{2}\left[g(t - T_0) + g(-t - T_0)\right]
 \end{aligned}
 $$

 -  O sistema é **não causal**, tendo-se em vista que 
 
 $$
 y(t) = S[g(t)]= \frac{1}{2}\left[g(t) + g(-t) \right].
 $$

Assim, por exemplo, $y(-1)$ depende da entrada futura $g(1)$.

iii. O que se pode dizer acerca da estabilidade de $S$? Justifique.

O sistema $S$ é **estável** pois dado um sinal limitado $g(t)$, a saída $g_p(t)$ também é limitada. De fato, temos que, se $|g(t)| < M_g < + \infty,$

 $$
 \begin{aligned}
 g_p(t) & = 0.5[g(t) + g(-t)], \text{ logo }\\
 |g_p(t)| & = 0.5|g(t) + g(-t)| \leq  0.5|g(t)| + 0.5|g(- t)| \leq M_g. \\
 \end{aligned}
 $$
 
 
 
iv. Considere os itens i. a iii. para o sistema $\tilde{S}$ tal que  $\tilde{S}[g(t)] = \frac{1}{2}\int^{+ \infty}_{-\infty} g(\tau) \left[\delta(t - \tau) + \delta(-t +  \tau) \right]d\tau$:

Neste caso, temos que o sistema é o sistema identidade, isto é, $S[g(t)] = g(t)$ e o sistema é portanto, linear, invariante no tempo, causal e estável.

### Questão 3.

Determine a potência média $P$ e a energia $E$ dos sinais abaixo:

i. $f(k) = \left(\frac{1}{2}\right)^{k}v(k), \quad \forall k \in \mathbb{Z},$


Temos que 
$$
E = \sum_{k = -\infty}^{+\infty}\left(\frac{1}{2}\right)^{2k}v(k) = \sum_{k = 0}^{+\infty}\left(\frac{1}{2}\right)^{2k} = \frac{1}{1 - \frac{1}{4}} =  \frac{4}{3}.
$$

Segue-se que $P =0$.

ii. $f(t) = e^{2t}v(-t), \quad \forall t \in \mathbb{R}$.
   Temos que 
$$
E = \int_{-\infty}^{0}e^{4t}dt = \int_{0}^{\infty}e^{-4t}dt = \left.-\frac{e^{-4t}}{4}\right|_0^{\infty} = \frac{1}{4}.
$$
Segue-se que $P =0$.
