### Exercício E1.2

Refaça o exemplo 1.2a para determinar a potência da senóide $ C \cos(\omega_0 t + \theta) $ calculando a média da energia do sinal em um período $ T_0 = 2 \pi \omega_0 $ (em vez de calcular a média em um intervalo infinitamente grande de tempo). Mostre também que a potência de um sinal $ C C x(t) = C_0 $ é $ C_0^2 $ e seu valor rms é $ C_0 $

### Exercício E1.6

Mostre que a exponencial de duração infinita $ e^{-at} $ não é nem um sinal de energia nem de potência para qualquer valor real de $ a $. Entretanto, se $ a $ for imaginário, ela é um sinal de potência por potência $ P_x = 1 $, independente do valor de $ a $.
#### Resolução:

Quando $ a $ é real, a energia $ E_x $ é zero, o que significa que o sinal não é um sinal de energia. E para a potência $ P_x $, não podemos definir uma janela de tempo finita $ T $
para calcular a média da potência, já que a integral se estende de $ -\infty $ a $ \infty $, e a exponencial de duração infinita nunca converge para um valor finito. Portanto, a potência também não é definida, indicando que o sinal não é um sinal de potência.

Entretanto, quando $ a $ é imaginário, $ e^{-at} = e^{-jwt} $. Nesse caso, a função se torna uma onda senoidal amortecida, e a energia $ E_x $ permanece zero. Porém, para calcular a potência $ P_x $, é possível utilizar uma janela de tempo $ T $ finita:

$ P_x = \lim _{T \to \infty} \frac{1}{T} \int_{\frac{T}{2}}^{\frac{-T}{2}} | e^{-jwt} | dt = \lim _{T \to \infty} \frac{1}{T} \int_{\frac{T}{2}}^{\frac{-T}{2}} 1 dt = 1$

Ou seja, a potência $ P_x $ é constante e igual a $ 1 $, independente do valor de $ b $.
