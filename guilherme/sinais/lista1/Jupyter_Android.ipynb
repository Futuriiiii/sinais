### Genéricas

Aqui discutimos algumas das operações mais básicas necessárias para a manipulação de expressões no SymPy.



#### subs

Uma das coisas mais comuns que você pode querer fazer com uma expressão matemática é a substituição. A substituição substitui todas as instâncias de algo em uma expressão por outra coisa. Isso é feito usando o método subs. Por exemplo

```python
expr = cos(x) + 1

expr.subs(x, y)
```

podemos também fazer uma substituição numérica, por $\pi$, por exemplo

```python
expr.subs(x, pi)
```

#### evalf

Para avaliar uma expressão numérica em um número de ponto flutuante, use *evalf.*



```python
expr = sqrt(8)
expr
```

```python
expr.evalf()
```

```python
2.82842712474619**2
```

```python
expr.evalf()**2
```

#### simplify

Uma das características mais úteis de um sistema de manipulação simbólica é a capacidade de simplificar expressões matemáticas. O SymPy tem dezenas de funções para realizar vários tipos de simplificação. Há também uma função geral chamada *simplify()* que tenta aplicar todas essas funções de forma inteligente para chegar à forma mais simples de uma expressão. 

Veja alguns exemplos:

```python
simplify(sin(x)**2 + cos(x)**2)
```

```python
simplify((x**3 + x**2 - x - 1)/(x**2 + 2*x + 1))
```

```python
limit((x**3 + x**2 - x - 1)/(x**2 + 2*x + 1),x, -1)
```

### Simplificação de polinômios ou funções racionais

#### expand

é uma das funções de simplificação mais comuns no SymPy. Embora tenha muitos escopos, por enquanto, vamos considerar sua função na expansão de expressões polinomiais. Por exemplo:

```python
expand((x + 1)**2)
```

```python
expand((x + 2)*(x - 3))
```

#### factor 

toma um polinômio e o transforma em fatores irredutíveis sobre os números racionais. Por exemplo:

```python
factor(x**3 - x**2 + x - 1)
```

#### collect

reúne poderes comuns de um termo em uma expressão. Por exemplo

```python
expr = x*y + x - 3 + 2*x**2 - z*x**2 + x**3
expr
```

```python
collected_expr = collect(expr, x)
collected_expr
```

#### cancel

tomará qualquer função racional e a colocará na forma canônica padrão, $\frac{p}{q}$, onde $p$ e $q$ são polinômios expandidos sem fatores comuns, e os principais coeficientes de $p$ e $q$ não têm denominadores (ou seja, são inteiros).

```python
cancel((x**2 + 2*x + 1)/(x**2 + x))
```

#### apart

Transformar funções racionais em frações parciais é bastante útil e existem vários métodos para isso (eliminação de frações, Heaviside etc). No sympy o método *apart* fornece esse resultado, como no exemplo abaixo.


```python
apart((3*x**2 + 9*x - 20)/(x**2 + x - 6))
```

```python
integrate((3*x**2 + 9*x - 20)/(x**2 + x - 6),x)
```

#### Exemplo 1.2

Determine a potência e o valor rms de

(a) $x(t)=C \cos \left(\omega_0 t+\theta\right)$

(b) $x(t)=C_1 \cos \left(\omega_1 t+\theta_1\right)+C_2 \cos \left(\omega_2 t+\theta_2\right) \quad \omega_1 \neq \omega_2$

(c) $x(t)=D e^{j \omega_0 t}$

```python
C, C1, C2, w0, w1, w2, th, th1, th2, t, T = symbols("C C1 C2 w0 w1  w2 th th1 th2 t T", nonzero=True)
```

(a)
$$
\begin{aligned}
P_x & =\lim _{T \rightarrow \infty} \frac{1}{T} \int_{-T / 2}^{T / 2} C^2 \cos ^2\left(\omega_0 t+\theta\right) d t=\lim _{T \rightarrow \infty} \frac{C^2}{2 T} \int_{-T / 2}^{T / 2}\left[1+\cos \left(2 \omega_0 t+2 \theta\right)\right] d t \\
& =\lim _{T \rightarrow \infty} \frac{C^2}{2 T} \int_{-T / 2}^{T / 2} d t+\lim _{T \rightarrow \infty} \frac{C^2}{2 T} \int_{-T / 2}^{T / 2} \cos \left(2 \omega_0 t+2 \theta\right) d t
\end{aligned}
$$

```python
Pa = (1/T)*integrate((C**2)*(cos(w0*t + th)**2),(t, -T/2, T/2))
simplify(Pa.subs(T, 2*pi/w0))
```

(b)

$$
\begin{aligned}
P_x= & \lim _{T \rightarrow \infty} \frac{1}{T} \int_{-T / 2}^{T / 2}\left[C_1 \cos \left(\omega_1 t+\theta_1\right)+C_2 \cos \left(\omega_2 t+\theta_2\right)\right]^2 d t \\
= & \lim _{T \rightarrow \infty} \frac{1}{T} \int_{-T / 2}^{T / 2} C_1^2 \cos ^2\left(\omega_1 t+\theta_1\right) d t+\lim _{T \rightarrow \infty} \frac{1}{T} \int_{-T / 2}^{T / 2} C_2^2 \cos ^2\left(\omega_2 t+\theta_2\right) d t \\
& +\lim _{T \rightarrow \infty} \frac{2 C_1 C_2}{T} \int_{-T / 2}^{T / 2} \cos \left(\omega_1 t+\theta_1\right) \cos \left(\omega_2 t+\theta_2\right) d t
\end{aligned}
$$

```python
Pb = (1/T)*integrate((C1*cos(w1*t + th1) + C2*cos(w2*t + th2))**2,(t, -T/2, T/2))
Pb = limit(Pb,T,oo)
simplify(Pb)
```
