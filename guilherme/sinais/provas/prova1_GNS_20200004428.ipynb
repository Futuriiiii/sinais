#### Questão 4

Memória e casualidade:

O seguinte sistema $ y(t) = x(-t) $ é com memória, pois a saída depende de valores em instantes de tempo diferentes da entrada. O sistema é não causal, pois depende de valores futuros da entrada.

Já o sistema $ y(t) = x(t) $ é sem memória, pois depende apenas de valores no mesmo instante de tempo da entrada. Esse sistema é não causal, pois não depende de valores futuros da entrada.

Variante no tempo ou invariante no tempo:

O seguinte sistema $ y(t) = e^{-t} \cdot x(t) $ é variante no tempo, pois a exponencial varia a saída de acordo com o tempo.

Por outro lado, o sistema $ y(t) = x(t + \tau) $ é invariante no tempo, pois a saída depende apenas do tempo, e variações em $ \tau $ correspondem apenas a deslocamentos temporais.

BIBO estável ou instável:

O sistema $ y(t) = \frac{1}{2}(x_1(t) + x_2(t)) $ é BIBO estável, pois se ambos $ x_1 $ e  $ x_2 $ forem estáveis, a saída também será.

Por outro lado, o sistema $ y(t) = k \cdot x(t) $ instável, pois o fator k varia a saída de acordo com o tempo, podendo resultar em valores ilimitados.
