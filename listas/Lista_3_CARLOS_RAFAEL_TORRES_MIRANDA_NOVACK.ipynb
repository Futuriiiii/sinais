## A Transformada de Laplace

Para um sinal $x(t)$, a transformada de Laplace $X(s)$ é definida por

$$
X(s) = \int_{-\infty}^{\infty} x(t) e^{-st} \, dt \tag{4.1}
$$

O sinal $x(t)$ é dito ser a **transformada inversa de Laplace** de $X(s)$. Pode ser mostrado que

$$
x(t) = \frac{1}{2\pi j} \int_{c - j \infty}^{c + j \infty} X(s) e^{st} \, ds \tag{4.2}
$$

onde $c$ é uma constante escolhida para garantir a convergência da integral da Eq. (4.1), como explicado posteriormente.

Esse par de equações é conhecido como **par da transformada de Laplace bilateral** (ou simplesmente **par de Laplace**), no qual $X(s)$ é a transformada direta de Laplace de $x(t)$ e $x(t)$ é a transformada inversa de Laplace de $X(s)$. Simbolicamente,

$$
X(s) = \mathcal{L}\{x(t)\} \quad \text{e} \quad x(t) = \mathcal{L}^{-1}[X(s)] \tag{4.3}
$$

Observe que

$$
\mathcal{L}^{-1}\{\mathcal{L}[x(t)]\} = x(t) \quad \text{e} \quad \mathcal{L}\{\mathcal{L}^{-1}[X(s)]\} = X(s)
$$

Também é prática comum utilizar uma seta bidirecional para indicar o par da transformada de Laplace, como mostrado a seguir:

$$
x(t) \iff X(s)
$$

A transformada de Laplace, definida desta forma, pode trabalhar com sinais que existem em todo o intervalo de tempo, de $-\infty$ a $\infty$ (sinais causais e não causais). Por essa razão, ela é chamada de **transformada de Laplace bilateral**.

### Questão 1: 

Prove que a ``transformada de Laplace é uma operação linear``, mostrando que o princípio da superposição é válido, isto é, se

$$
x_1(t) \Longleftrightarrow X_1(s) \quad \mathrm{e} \quad x_2(t) \Longleftrightarrow X_2(s)
$$

então

$$
a_1 x_1(t)+a_2 x_2(t) \Longleftrightarrow a_1 X_1(s)+a_2 X_2(s)
$$


## RESPOSTA Q.1

A transformada de Laplace de uma função $ x(t) $ é dada por:

$$
\mathcal{L}\{x(t)\} = X(s) = \int_0^\infty x(t) e^{-st} \, dt
$$

Queremos mostrar que, se $ x_1(t) \Longleftrightarrow X_1(s) $ e $ x_2(t) \Longleftrightarrow X_2(s) $, então

$$
a_1 x_1(t) + a_2 x_2(t) \Longleftrightarrow a_1 X_1(s) + a_2 X_2(s)
$$

### Prova

Seja a função $ x(t) = a_1 x_1(t) + a_2 x_2(t) $, onde $ a_1 $ e $ a_2 $ são constantes. Vamos calcular a transformada de Laplace de $ x(t) $:

$$
\mathcal{L}\{a_1 x_1(t) + a_2 x_2(t)\} = \int_0^\infty \left( a_1 x_1(t) + a_2 x_2(t) \right) e^{-st} \, dt
$$

Usando a propriedade distributiva da integral:

$$
\mathcal{L}\{a_1 x_1(t) + a_2 x_2(t)\} = a_1 \int_0^\infty x_1(t) e^{-st} \, dt + a_2 \int_0^\infty x_2(t) e^{-st} \, dt
$$

Agora, aplicando a definição da transformada de Laplace para $ x_1(t) $ e $ x_2(t) $:

$$
\mathcal{L}\{a_1 x_1(t) + a_2 x_2(t)\} = a_1 X_1(s) + a_2 X_2(s)
$$

Assim, provamos que:

$$
a_1 x_1(t) + a_2 x_2(t) \Longleftrightarrow a_1 X_1(s) + a_2 X_2(s)
$$

Portanto, a **transformada de Laplace** é uma **operação linear**.

### Questão 2: 
Para o sinal $x(t)=e^{-a t} u(t)$, determine sua transformada de Laplace $X(s)$ e sua **região de convergência (RDC)**, também chamada de região de existência, dada pelo o conjunto de valores de $s$ (a região no plano complexo) para os quais a integral da Eq. (4.1) converge.

## RESPOSTA Q.2

```python
from sympy import symbols, exp, laplace_transform, Heaviside

# Definindo as variáveis
t, s, a = symbols('t s a', real=True)

# Sinal x(t) = e^(-a*t) * u(t), onde u(t) é a função degrau unitário (Heaviside)
x_t = exp(-a*t) * Heaviside(t)

# Calculando a transformada de Laplace
X_s, convergence, _ = laplace_transform(x_t, t, s)

X_s, convergence
```

1. **Inversão no tempo**: A inversão no tempo, representada por $ u(-t) $, altera o suporte do sinal, concentrando-o na região negativa do tempo $ t \leq 0 $, enquanto a transformada de Laplace é comumente definida para $ t \geq 0 $.

2. **Transformada de Laplace nula**: Para o sinal $ x(t) = -e^{-a t} u(-t) $, a transformada de Laplace resulta em $ X(s) = 0 $, pois o sinal não tem contribuição na região $ t \geq 0 $, que é a região de integração usual da transformada de Laplace.

3. **Diferença fundamental**: O sinal $ x(t) = e^{-a t} u(t) $ tem suporte na região positiva $ t \geq 0 $, o que resulta em uma transformada de Laplace $ X(s) = \frac{1}{a+s} $ com uma região de convergência bem definida, ao contrário do caso invertido no tempo.

## RESPOSTA Q.8

### Questão 8: 

A transformada de Laplace de uma função $ x(t) $ é dada por:

$$
\mathcal{L}\{x(t)\} = X(s) = \int_0^{\infty} x(t) e^{-st} dt
$$

Agora, considere o sinal deslocado $ x(t - t_0) $. A transformada de Laplace deste sinal é:

$$
\mathcal{L}\{x(t - t_0)\} = \int_0^{\infty} x(t - t_0) e^{-st} dt
$$

Realizando a substituição de variáveis, seja $ u = t - t_0 $, então $ du = dt $. Quando $ t = 0 $, $ u = -t_0 $, e quando $ t \to \infty $, $ u \to \infty $. Assim, a integral torna-se:

$$
\int_{-t_0}^{\infty} x(u) e^{-s(u + t_0)} du
$$

Isso pode ser reescrito como:

$$
e^{-s t_0} \int_{-t_0}^{\infty} x(u) e^{-su} du
$$

Sabendo que $ x(u) = 0 $ para $ u < 0 $, podemos ajustar os limites da integral para:

$$
e^{-s t_0} \int_0^{\infty} x(u) e^{-su} du
$$

Que é simplesmente:

$$
e^{-s t_0} X(s)
$$

Portanto, mostramos que:

$$
\mathcal{L}\{x(t - t_0)\} = X(s) e^{-s t_0}
$$
